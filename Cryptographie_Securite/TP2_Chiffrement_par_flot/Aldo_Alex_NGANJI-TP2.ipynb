```python
def stringToBinary(msg):
    msg_bin = ""
    for i in bytearray(msg, encoding ='ascii') :
        msg_bin = msg_bin + format(i, '08b')
    return msg_bin
def binaryToString(binary):
    msg = ""
    for i in range(0, len(binary), 8):
        byte_int = int(binary[i:i+8], 2)
        byte_char = chr(byte_int)
        msg = msg + byte_char
    return msg
    
print("En binaire :", stringToBinary("message en clair"))
print("En ascii :",binaryToString(stringToBinary("message en clair")))
```

```python
def chiffrementVernam(msgBinaire, clef):
    chiffre = ""
    for i in range(len(msgBinaire)):
        bit_msg = int(msgBinaire[i])
        bit_clef = int(clef[i])
        chiffre += str(bit_msg ^ bit_clef)
    return chiffre

try:
    assert chiffrementVernam(stringToBinary("vernam"), "110011001100110011001100110011001100110011001100") == "101110101010100110111110101000101010110110100001"
    print("chiffrementVernam : OK")
except:
    print("chiffrementVernam : ERREUR")
```

```python
def dechiffrementVernam(msgChiffre, clef):
    clair_bin = ""
    for i in range(len(msgChiffre)):
        bit_chiffre = int(msgChiffre[i])
        bit_clef = int(clef[i])
        clair_bin += str(bit_chiffre ^ bit_clef)
    return clair_bin

chiffre = "1010001010101101101010011011111010111000"
clef = "1100110011001100110011001100110011001100"

clair_binaire = dechiffrementVernam(chiffre, clef)
msg_clair = binaryToString(clair_binaire)

print("Message clair binaire :", clair_binaire)
print("Message clair ASCII :", msg_clair)
```

```python
def etatSuivant(etat, coeff):
    bit_sortie = etat[0]
    nouveau_bit = 0
    for idx in coeff:
        nouveau_bit ^= etat[idx]
    nouvel_etat = etat[1:] + [nouveau_bit]
    return nouvel_etat, bit_sortie

try:
    assert etatSuivant([1,0,0,1], [0,2,3]) == ([0, 0, 1, 0], 1)
    assert etatSuivant(etatSuivant([1,0,0,1], [0,2,3])[0], [0,2,3]) == ([0, 1, 0, 1], 0)
    print("etatSuivant : OK")
except:
    print("etatSuivant : ERREUR")
```
