```python
def code(s):
    tab=[]
    for i in s:
        tab.append(ord(i)-65)
    return tab

print(code('ABCYZ'))
```

```python
def decode(tab):
    s = ""
    for n in tab:
        s = s + chr(n + 65)
    return s
```

```python
def cesar(chaine, n):
    liste_nombres = code(chaine)
    
    liste_chiffree = []
    for num in liste_nombres:
        num_c = (num + n) % 26
        liste_chiffree.append(num_c)
        
    return decode(liste_chiffree)
```

```python
message_original = "BONJOUR"
message_chiffre = cesar(message_original, 3)

print("Message clair :", message_original)
print("Message chiffré :", message_chiffre)
```
