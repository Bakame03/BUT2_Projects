```python
def expo_mod(a, r, m):
    p = 1
    a = a % m
    while r > 0:
        if r % 2 == 1:
            p = (p * a) % m
            r = (r - 1) // 2
        else:
            r = r // 2
        a = (a * a) % m
    return p

expo_mod(13, 5, 8)
```
