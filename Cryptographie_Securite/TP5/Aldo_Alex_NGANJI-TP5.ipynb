```python
def puissance(a, x, n):
    """methode récursive, rapide de calcul de a^x mod n
    Attention au cas x == 0"""
    if x == 0:
        return 1
    elif x == 1:
        return a
    elif x % 2 == 0:
        return puissance(a**2 % n, x // 2, n) % n
    elif x != 1:
        return a * puissance(a**2 % n, (x-1) // 2, n) % n


def cycle(a, p):
    """retourne a**0, a**1, a**2, ...a**(p-1) modulo p """
    resultat = []
    for d in range(p-1):
        resultat.append(puissance(a, d, p))
    return resultat


for a in [31, 2, 3, 4, 5]:
    print(a, cycle(a, 31))
```

```python
def primitive(a, p):
    for d in range(1, p - 1):
        if puissance(a, d, p) == 1:
            return False
    return True

primitive(3, 31)
```

```python
def racine_primitive(p):
    for a in range(2, p):
        if primitive(a, p):
            return a

def liste_racines_primitives(p):
    resultat = []
    for a in range(2, p):
        if primitive(a, p):
            resultat.append(a)
    return resultat
```

```python
from sympy import randprime
from random import randint

def clefs(k):
    p = randprime(2**k, 2**(k+1))
    g = racine_primitive(p)
    a = randint(1, p - 2)
    A = puissance(g, a, p)
    public_key = (p, g, A)
    private_key = a
    return public_key, private_key
```

```python
def chiffrer(m, public_key):
    (p, g, A) = public_key
    b = randint(0, p - 2)
    return m * puissance(A, b, p) % p, puissance(g, b, p)
```

```python
def dechiffrer(c, B, public_key, private_key):
    (p, g, A) = public_key
    return c * puissance(puissance(B, private_key, p), p - 2, p) % p
```

```python
def encode(texte):
    chaine = ""
    for car in texte:
        code = str(ord(car))
        while len(code) < 3:
            code = "0" + code
        chaine = chaine + code
    return int(chaine)

def decode(nombre):
    texte = ""
    while nombre > 0:
        texte = chr(nombre % 1000) + texte
        nombre = nombre // 1000
    return texte
```
